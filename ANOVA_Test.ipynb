```r
install.packages("tidyverse")
library(tidyverse)
set.seed(42)
```

```r
# Generating Dataset With 4 Treatment Groups
anova_data <- tibble(
  patient_ID = 1:120,
  drug_group = factor(rep(c("Drug_A" , "Drug_B" , "Drug_C" , "Placebo") , each = 30)),
  bp_drop = c(rnorm(30, mean = 18, sd = 4) , # Drug A Works Great
  rnorm(30, mean = 12, sd = 4), # Drug B Works Okay
  rnorm(30, mean = 5, sd = 4), # Drug C Is Weak
  rnorm(30, mean = 2, sd = 4) # Placebo)
  )
)
```

```r
# Performing NOVA (Analysis Of Variance)
anova_result <- aov(bp_drop ~ drug_group, data = anova_data)
summary(anova_result)
```

```r
TukeyHSD(anova_result)
```

### Project Summary:

1. Purpose:
The Objective Of This Analysis Was To Evaluate The Effectiveness Of Three Different Drug Groups (Drug_A, Drug_B, Drug_C) Compared To A Control Group (Placebo)on Lowering Blood Pressure (bp_drop).

2. ANOVA Results:
The One-Way ANOVA Test Revealed A Highly Significant Difference Between The Treatment Groups (F(3, 116) = 87.45, p < 2e-16). This Indicates That The Choice Of Drug Significantly Impacts Blood Pressure Reduction.

3. Post-Hoc Tukey HSD Results:
The Pairwise Comparisons Showed That All groups Are Statistically Significantly Different From One Another (All Adjusted p-values < 0.05).
- Drug_A Showed The Highest Efficiency In Reducing Blood Pressure.
- Drug_B Performed Moderately Well, Significantly Better That Drug_C And Placebo, But Less Than Drug_A.
- Drug_C Had A Weak Effect, But It Was Still Statistically Better That The Placebo Group.

Conclusion:
The Statistical Hierarchy Of Effectiveness Is: Drug_A > Drug_B > Drug_C > Placebo.
