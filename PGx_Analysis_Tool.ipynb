```python
# ===================================
# 1. Data Initialization and Data Loading
# ===================================

import pandas as pd
try:
  df = pd.read_csv("drugsdata.csv")
  df.loc[df['drug_name'].str.contains('Clopidogrel', case=False), 'gene'] = 'CYP2C19'
  df.loc[df['drug_name'].str.contains('Aspirin', case=False), 'gene'] = 'PTGS1'
  df.to_csv('drugsdata.csv', index=False)
  print("Successfully loaded and updated the Pharmacogenomics Database.")
except FileNotFoundError:
  print("Error: 'drugsdata.csv' File not found. Please check the path.")
print("-" * 50)

# ===================================
# 2. Clinical Decision Support System
# ===================================

def lookup_drug(drug_name, dataframe):
  result = dataframe[dataframe['drug_name'].str.contains(drug_name, case=False, na=False)]
  if not result.empty:
    row = result.iloc[0]
    print("=== Genetic Recommendation Found ===")
    print(f"Drug:             {row['drug_name']}")
    print(f"Associated Gene:  {row['gene']}")
    print(f"Medical Advice:   {row['recommendation']}")
  else:
    print(f"No Data Found For '{drug_name}'.")
print("-" * 50)

# ===================================
# 3. The Interaction Loop
# ===================================

while True:
  search_drug = input("Enter drug name (or type exit to quit):").strip()
  if search_drug.lower() == 'exit':
    print("Closing the Clinical Decision Support Sestem.")
    break
  if search_drug:
    lookup_drug(search_drug, df)
```
