```r
# Create A Simple Dataset Of Patients
patient_data <- data.frame(
    patient_ID = c(1, 2, 3, 4, 5),
    name = c("Sara" , "John" , "Amal" , "Mike" , "Elena"),
    age = c(25, 45, 30, 60, 52),
    blood_sugar = c(95, 140, 110, 210, 135)
)
print(patient_data)
```

```r
# Filter Patients With Blood Sugar Greater Than 130
high_sugar_patients <- patient_data[patient_data$blood_sugar > 130, ]
print(high_sugar_patients)
```

```r
# Create A Simple Bar Plot For Blood Sugar Levels
barplot(
  patient_data$blood_sugar,
  names.arg = patient_data$name,
  col = "blue" ,
  main = "Patients Blood Sugar Levels" ,
  xlab = "Patient Name" ,
  ylab = "Blood Sugar (mg/dl)"
)

# Add A Horizontal Line At 130 To Show The Danger Threshold
abline(h = 130, col = "red" , lty = 2 , lwd = 2)
```
