```r
install.packages("tidyverse")
library(tidyverse)
set.seed(42)
```

```r
# Generating Dirty Clinical Data
dirty_data <- tibble(
  patient_ID = 1:50 ,
  heart_rate = sample(60:100, 50, replace = TRUE)
)
dirty_data$heart_rate[c(5, 15, 25)] <- NA
dirty_data$heart_rate[c(10, 40)] <- 250
```

```r
# Cleaning The Data
clean_data <- dirty_data %>%
  mutate(heart_rate = ifelse (is.na(heart_rate), mean(heart_rate, na.rm = TRUE), heart_rate))%>%
  filter(heart_rate <= 200)
head(clean_data)
```
