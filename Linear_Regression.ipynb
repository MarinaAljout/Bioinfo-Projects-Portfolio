```r
install.packages("tidyverse")
library(tidyverse)
set.seed(42)
```

```r
clinical_data_regression <- tibble(
  patient_ID = 1:100,
  age = sample(20:80, 100, replace = TRUE) ,
  weight = sample(50:110, 100, replace = TRUE) ,
  bp_observed = 90 + (0.4 * age) + (0.3 * weight) + rnorm(100, mean = 0, sd = 5)
)
head(clinical_data_regression)
```

```r
# Building The Linear Regression Model
model <- lm(bp_observed ~ age + weight, data = clinical_data_regression)
summary(model)
```

# Model Interpretation:
