```r
install.packages("tidyverse")
library(tidyverse)
set.seed(42)
clinical_trial <- tibble(
    patient_ID = 1:100,
    bp_before = sample(140:180, 100, replace = TRUE) ,
    bp_after = bp_before - round(rnorm(100, mean = 15, sd = 10))
)

head(clinical_trial)
```

```r
# Performing Paired t-test
ttest_result <- t.test(clinical_trial$bp_before , clinical_trial$bp_after , paired = TRUE)
print(ttest_result)
```

* **'t = 15.555'** ==> The durg's effect is huge (far from 0).

* **'df = 99'** ==> Number of patients minus 1.

* **'p-value < 2.2e-16'** ==> Extremely small (less than 0.05); proves the drug works and it is NOT luck.

* **'95% confidence interval'** ==> Future patients will see a blood pressure drop between 12.43 and 16.06 points.

* **'mean difference = 14.25'** ==> On average, patients' blood pressure dropped by 14.25 points.
