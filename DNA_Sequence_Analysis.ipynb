```python
# ===================================
# 1. Data Initialization & Inputs
# ===================================

dna_seq = "GGCCATGCGATTACGATCGACTGAACCA"
start_motif = "ATGC"
end_motif = "TGA"

# ===================================
# 2. Core analysis Function
# ===================================

def analyze_dna_segment(dna_sequence, start_motif, end_motif):
  sequence = dna_sequence.upper()
  start_index = sequence.find(start_motif.upper())
  end_index = sequence.find(end_motif.upper())
  if start_index == -1 or end_index == -1:
    print("Warning: Start Or End Motif Not Found In The Sequence!")
    return None
  extracted_segment = sequence[start_index : end_index + len(end_motif)]
  segment_length = len(extracted_segment)
  is_valid = segment_length % 3 == 0
  status = "Yes (Valid Coding Logic)" if is_valid else "No (Potential Framshift Mutation)"
  report = {
      "Segment": extracted_segment,
      "Length" : segment_length,
      "Is_Valid" : status
  }
  return report

# ===================================
# 3. Pipeline Execution & Output
# ===================================

analysis_result = analyze_dna_segment(dna_seq, start_motif, end_motif)
if analysis_result:
  print("=== DNA ANALYSIS REPORT ===")
  print(f"Extracted Segment: {analysis_result['Segment']}")
  print(f"Sequence Length:   {analysis_result['Length']}")
  print(f"Is Valid Codon?    {analysis_result['Is_Valid']}")
```
