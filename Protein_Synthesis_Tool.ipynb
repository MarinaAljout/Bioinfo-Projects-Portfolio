```python
# ===================================
# 1. Reference Data & Genetic Code
# ===================================

genetic_code = {
    'AUG': {'name': 'Methionine', 'weight': 149.2},
    'GCC': {'name': 'Alanine', 'weight': 89.1},
    'AUU': {'name': 'Isoleucine', 'weight': 131.2},
    'GUA': {'name': 'Valine', 'weight': 117.1},
    'GGC': {'name': 'Glycine', 'weight': 75.1},
    'CGC': {'name': 'Arginine', 'weight': 174.2},
    'AAA': {'name': 'Lysine', 'weight': 146.2},
    'CCG': {'name': 'Proline', 'weight': 115.1},
    'UCU': {'name': 'Serine', 'weight': 105.1},
    'ACC': {'name': 'Threonine', 'weight': 119.1},
    'UGG': {'name': 'Tryptophan', 'weight': 204.2},
    'UAA': {'name': 'STOP', 'weight': 0},
    'UAG': {'name': 'STOP', 'weight': 0},
    'UGA': {'name': 'STOP', 'weight': 0}
}
dna_seq = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTCCCTAA"

# ===================================
# 2. Core Translation Function
# ===================================

def translate_mrna(mrna_seq):
  total_weight = 0
  amino_acids_chain = []
  for i in range(0, len(mrna_seq), 3):
    codon = mrna_seq[i:i+3]
    if len(codon) < 3:
      break
    if codon in genetic_code:
      amino_name = genetic_code[codon]['name']
      amino_weight = genetic_code[codon]['weight']
      if amino_name == "STOP":
        print("Stop Codon Reached.")
        break
      print(f" Adding: {amino_name} (Weight: {amino_weight} Da)")
      total_weight += amino_weight
      amino_acids_chain.append(amino_name)
    else:
      print(f"Codon {codon}: Unknown")
  return total_weight, amino_acids_chain

# ===================================
# 3. Mutation Simulation Function
# ===================================

def simulate_mutation(dna_seq, index, new_base):
  dna_list = list(dna_seq)
  dna_list[index] = new_base.upper()
  return "".join(dna_list)

# ===================================
# 4. Pipeline Execution & Comparison
# ===================================

print("=== ANALYZING ORIGINAL SEQUENCE ===")
mrna_normal = dna_seq.replace('T', 'U')
print(f"mRNA: {mrna_normal}")
weight_normal, chain_normal = translate_mrna(mrna_normal)
print(f"Total Protein Molecular Weight: {weight_normal:.2f} Da")
print("-" * 50)

print("=== SIMULATING & ANALYZING MUTATION ===")
mutated_dna = simulate_mutation(dna_seq, 12, 'T')
print(f"Original DNA: {dna_seq}")
print(f"Mutated DNA: {mutated_dna}")
mrna_mutated = mutated_dna.replace('T', 'U')
print(f"Mutated mRNA: {mrna_mutated}")
weight_mutated, chain_mutated = translate_mrna(mrna_mutated)
print(f"Total Protein Molecular Weight: {weight_mutated:.2f} Da")
```
